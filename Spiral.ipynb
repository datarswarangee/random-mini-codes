Inward Spiral Drawing GIF

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from google.colab import files

# SETTINGS
FIG_SIZE = 6

TURNS = 5

N_POINTS = 1200

N_FRAMES = 120

LINE_WIDTH = 2.5

FPS = 30

MAX_RADIUS = 1.0

# CREATE SPIRAL

# Angle from center outward
theta = np.linspace(0, 2 * np.pi * TURNS, N_POINTS)

# Radius increases gradually from center outward
radius = MAX_RADIUS * theta / theta.max()

# Convert polar coordinates to Cartesian coordinates
x = radius * np.cos(theta)
y = radius * np.sin(theta)


# CREATE FIGURE
fig, ax = plt.subplots(figsize=(FIG_SIZE, FIG_SIZE))

fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# Remove axes
ax.set_xlim(-1.05, 1.05)
ax.set_ylim(-1.05, 1.05)
ax.set_aspect("equal")
ax.axis("off")

# Create empty black spiral
line, = ax.plot([], [], color="black", linewidth=LINE_WIDTH)

# INITIALIZE ANIMATION
def init():
    line.set_data([], [])
    return line,

# ANIMATION FUNCTION
def animate(frame):

    # Progress from 0 → 1
    progress = frame / (N_FRAMES - 1)

    # Number of spiral points shown
    current_points = int(progress * N_POINTS)

    # Draw spiral from center outward
    line.set_data(
        x[:current_points],
        y[:current_points]
    )

    return line,


# CREATE ANIMATION
animation = FuncAnimation(
    fig,
    animate,
    init_func=init,
    frames=N_FRAMES,
    interval=1000 / FPS,
    blit=True
)

# SAVE AS GIF
output_file = "inward_to_outward_spiral.gif"

animation.save(
    output_file,
    writer=PillowWriter(fps=FPS)
)

plt.close(fig)

print("GIF created successfully!")
print(f"File: {output_file}")

# DOWNLOAD GIF
files.download(output_file)
```
